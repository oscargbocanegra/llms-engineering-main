## And now for some fun - an adversarial conversation between Chatbots..

You're already familar with prompts being organized into lists like:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

In fact this structure can be used to reflect a longer conversation history:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

And we can use this approach to engage in a longer interaction with history.

````python
# Let's create a conversation between Gemini and Ollama
print(NVIDIA_MODEL)
print(OLLAMA_MODEL_GEMMA)
````

````python
ollama_system = "You are a very argumentative chatbot; you disagree with everything in the conversation and question everything in a sarcastic way"

nvidia_system = "You are a very polite and courteous chatbot. You try to agree with everything the other person says or find common ground. \
If the other person argues, you try to calm them down and keep chatting" 
````

````python
def call_ollama_conversation():
    messages = [{"role": "system", "content": ollama_system,}
    ]

    for ollama_msg, nvidia_msg in zip(ollama_messages, nvidia_messages,):
        messages.append({"role": "assistant", "content": ollama_msg,})
        messages.append({"role": "user","content": nvidia_msg,})

    completion = call_model(
        client=ollama_client,
        model=OLLAMA_MODEL_GEMMA,
        messages=messages,
        use_stream=True,
        display_output=False,
    )

    return completion
````

````python
def call_nvidia_conversation():
    messages = [{"role": "system", "content": nvidia_system, }]

    for ollama_msg, nvidia_msg in zip(ollama_messages, nvidia_messages,):
        messages.append({"role": "assistant","content": ollama_msg,})
        messages.append({"role": "user","content": nvidia_msg,})

    messages.append({"role": "user","content": ollama_messages[-1],})

    completion = call_model(
        client=nvidia_client,
        model=NVIDIA_MODEL,
        messages=messages,
        use_stream=True,
        display_output=False,
    )

    return completion
````

````python
ollama_messages = ["Hi"]
nvidia_messages = ["Hi"]
````

````python
ollama_response = call_ollama_conversation()
````

````python
nvidia_response = call_nvidia_conversation()
````

````python
# Run the conversation for 5 rounds
ollama_messages = ["Hi"]
nvidia_messages = ["Hi"]

for i in range(5):
    print(f"\n--- Round {i + 1}: Ollama ---")
    ollama_next = call_ollama_conversation()
    ollama_messages.append(ollama_next)

    print(f"\n--- Round {i + 1}: NVIDIA ---")
    nvidia_next = call_nvidia_conversation()
    nvidia_messages.append(nvidia_next)
````
